```python
num1  = float(input("1st number: "))
num2 = float(input("2nd number: "))
print("Select operator:")
print("  + for addition")
print("  - for subtraction")
print("  * for multiplication")
print("  / for division")
op = input("Enter operator: ")

if op == '+':
    result = num1 + num2
elif op == '-':
    result = num1 - num2
elif op == '*':
    result = num1 * num2
elif op == '/':
    if num2 == 0:
        print("Error: Can't divide by zero")
    else:
        result = num1 / num2
        print("Result:", result)
else:
    print("Invalid operation")

if op in ['+', '-', '*']:
    print("Result:", result)
```
