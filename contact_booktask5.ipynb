```python
import os

name_list = []
phone_list = []
email_list = []
address_list = []

choice = 0

while choice != 6:
    os.system('clear')
    print("--- Contact Book ---")
    print("1.Add  2.View  3.Search  4.Update  5.Delete  6.Exit")
    choice = int(input("Enter number: "))

    if choice == 1:
        n = input("Name: ")
        p = input("Phone: ")
        e = input("Email: ")
        a = input("Address: ")
        name_list.append(n)
        phone_list.append(p)
        email_list.append(e)
        address_list.append(a)
        print("Contact added!")

    elif choice == 2:
        if len(name_list) == 0:
            print("No contacts saved yet.")
        else:
            for i in range(len(name_list)):
                print(i+1, name_list[i], "-", phone_list[i])

    elif choice == 3:
        search = input("Enter name or phone: ")
        found = False
        for i in range(len(name_list)):
            if search == name_list[i] or search == phone_list[i]:
                print("Name:", name_list[i])
                print("Phone:", phone_list[i])
                print("Email:", email_list[i])
                print("Address:", address_list[i])
                found = True
        if found == False:
            print("Contact not found.")

    elif choice == 4:
        search = input("Enter name to update: ")
        found = False
        for i in range(len(name_list)):
            if search == name_list[i]:
                phone_list[i] = input("New phone: ")
                email_list[i] = input("New email: ")
                address_list[i] = input("New address: ")
                print("Updated!")
                found = True
        if found == False:
            print("Contact not found.")

    elif choice == 5:
        search = input("Enter name to delete: ")
        found = False
        for i in range(len(name_list)):
            if search == name_list[i]:
                name_list.pop(i)
                phone_list.pop(i)
                email_list.pop(i)
                address_list.pop(i)
                print("Deleted!")
                found = True
                break
        if found == False:
            print("Contact not found.")

    elif choice == 6:
        print("Bye!")

    else:
        print("Enter number between 1 and 6")
```
