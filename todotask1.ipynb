```python
tasks = []

def add_task():
    t = input("Enter task: ")
    tasks.append([t, False])
    print("Task added!")

def view_tasks():
    if len(tasks) == 0:
        print("No tasks!")
        return
    for i, t in enumerate(tasks, 1):
        status = "Done" if t[1] else "Pending"
        print(f"{i}. {t[0]} - {status}")

def update_task():
    view_tasks()
    num = int(input("Enter task number to update: "))
    new_name = input("Enter new task name: ")
    tasks[num - 1][0] = new_name
    print("Task updated!")

def mark_complete():
    view_tasks()
    num = int(input("Enter task number to mark as completed: "))
    tasks[num - 1][1] = True
    print("Task marked as completed!")

def delete_task():
    view_tasks()
    n = int(input("Enter task number to delete: ")) - 1
    tasks.pop(n)
    print("Deleted!")

while True:
    print("\n1. Add  2. View  3. Update  4. Mark as Complete  5. Delete")
    ch = input("Choice: ")
    if ch == "1": add_task()
    elif ch == "2": view_tasks()
    elif ch == "3": update_task()
    elif ch == "4": mark_complete()
    elif ch == "5": delete_task()
```
