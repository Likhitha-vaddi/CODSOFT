```python
import random
```

```python
wins = 0
losses = 0

print("Welcome to Rock - Paper - Scissors!")

beat = {"rock": "scissors", "scissors": "paper", "paper": "rock"}

while True:
    user = input("\nYour choice (rock/paper/scissors): ").lower()

    if user not in beat:
        print("Wrong input! Try rock, paper or scissors.")
        continue

    computer = random.choice(list(beat.keys()))
    print(f"Your pick : {user}  |  Computer pick : {computer}")

    if user == computer:
        print("It's a tie!")
    elif beat[user] == computer:
        print("You win!")
        wins += 1
    else:
        print("Computer wins!")
        losses += 1

    print("Score --> Wins:", wins, "| Losses:", losses)

    if input("Play again? (yes/no): ").lower() != "yes":
        print("Thanks for playing! Final Score --> Wins:", wins, "| Losses:", losses)
        break
```
